```python
import pandas as pd
from pathlib import Path
from src.utils.paths import CONFIG_DIR
import yaml
import numpy as np
```

```python
from src.utils.paths import PROCESSED_DIR
```

```python
path = PROCESSED_DIR / "alipur"
df = pd.read_csv(path)
df['Timestamp'] = pd.to_datetime(df['Timestamp'])
df = df.set_index('Timestamp')
df.rename_axis(None, inplace=True)
df.head()
```

```python
with open(str(CONFIG_DIR / "config.yaml")) as f:
    config = yaml.safe_load(f)
```

```python
instances = df.shape[0] - 3 - 1 + 1
tr_instances = int(instances * config["split"]["train"])
vl_instances = int(instances * config["split"]["val"])

print(tr_instances, vl_instances)
```

```python
train_rows = df.iloc[: tr_instances + 3 + 1 - 1] # Combinations
vmin = np.min(train_rows, axis=0)
vmax = np.max(train_rows, axis=0)

scale = np.where(vmax - vmin == 0, 1.0, vmax - vmin)
scaled = (train_rows - vmin) / scale

print(scaled)
```
